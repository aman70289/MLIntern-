### Load Titanic Dataset
The Titanic dataset is a classic dataset often used for classification tasks. It contains information about passengers on the Titanic, including their demographics, ticket details, and whether they survived the disaster. We will load it directly from a public URL.

```python
import pandas as pd

# URL for the Titanic dataset (Kaggle)
titanic_url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

# Load the dataset into a pandas DataFrame
df_titanic = pd.read_csv(titanic_url)

# Display the first 5 rows of the DataFrame
display(df_titanic.head())
```

### Exploratory Data Analysis (EDA)
We will start by examining the basic structure of the dataset, checking data types, and identifying any missing values.

```python
# Display general information about the DataFrame
df_titanic.info()
```

```python
# Display descriptive statistics for numerical columns
display(df_titanic.describe())
```

```python
# Check for missing values
display(df_titanic.isnull().sum())
```

### Data Cleaning: Handling Missing Values and Irrelevant Columns
Based on the initial EDA, we will address missing values in 'Age' and 'Embarked' and drop 'Cabin' due to too many missing entries. We will also drop 'PassengerId', 'Name', and 'Ticket' as they are not directly useful for predictive modeling.

```python
# Impute 'Age' with the median
df_titanic['Age'].fillna(df_titanic['Age'].median(), inplace=True)

# Impute 'Embarked' with the mode
df_titanic['Embarked'].fillna(df_titanic['Embarked'].mode()[0], inplace=True)

# Drop 'Cabin' column due to high number of missing values
df_titanic.drop('Cabin', axis=1, inplace=True)

# Drop irrelevant columns: 'PassengerId', 'Name', 'Ticket'
df_titanic.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)

print("Missing values after handling:")
display(df_titanic.isnull().sum())
```

### Feature Engineering: Encoding Categorical Variables
Now we will convert categorical features ('Sex', 'Embarked') into numerical format using one-hot encoding, which is suitable for most machine learning algorithms.

```python
# Convert 'Sex' column to numerical (0 for male, 1 for female)
df_titanic['Sex'] = df_titanic['Sex'].map({'male': 0, 'female': 1})

# One-hot encode 'Embarked' column
df_titanic = pd.get_dummies(df_titanic, columns=['Embarked'], drop_first=True)

print("DataFrame after encoding categorical variables:")
display(df_titanic.head())
```

### Normalization / Standardization
Scaling numerical features is an important preprocessing step. We will use `StandardScaler` to transform features like 'Age', 'Fare', 'SibSp', and 'Parch' so they have a mean of 0 and a standard deviation of 1.

```python
from sklearn.preprocessing import StandardScaler

# Define numerical features to scale
numerical_features = ['Age', 'Fare', 'SibSp', 'Parch']

# Initialize StandardScaler
scaler = StandardScaler()

# Apply StandardScaler to the selected numerical features
df_titanic[numerical_features] = scaler.fit_transform(df_titanic[numerical_features])

print("DataFrame after scaling numerical features:")
display(df_titanic.head())
```

### In-depth Exploratory Data Analysis (EDA) with Visualizations
Let's visualize the relationships between various features and our target variable, 'Survived'. This will provide deeper insights into the data and help in potential further feature engineering or model selection.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style of the plots
sns.set_style("whitegrid")

# 1. Survival by Sex
plt.figure(figsize=(6, 4))
sns.barplot(x='Sex', y='Survived', data=df_titanic, palette='viridis')
plt.title('Survival Rate by Sex')
plt.xticks(ticks=[0, 1], labels=['Male', 'Female'])
plt.ylabel('Survival Rate')
plt.show()

# 2. Survival by Pclass
plt.figure(figsize=(7, 5))
sns.barplot(x='Pclass', y='Survived', data=df_titanic, palette='magma')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.show()

# 3. Survival by Age (Distribution)
plt.figure(figsize=(10, 6))
sns.histplot(df_titanic[df_titanic['Survived'] == 1]['Age'], color='green', kde=True, label='Survived')
sns.histplot(df_titanic[df_titanic['Survived'] == 0]['Age'], color='red', kde=True, label='Not Survived')
plt.title('Age Distribution of Survivors vs. Non-Survivors')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend()
plt.show()

# 4. Survival by SibSp (Siblings/Spouses Aboard)
plt.figure(figsize=(8, 5))
sns.barplot(x='SibSp', y='Survived', data=df_titanic, palette='cividis')
plt.title('Survival Rate by Number of Siblings/Spouses Aboard')
plt.ylabel('Survival Rate')
plt.show()

# 5. Survival by Parch (Parents/Children Aboard)
plt.figure(figsize=(8, 5))
sns.barplot(x='Parch', y='Survived', data=df_titanic, palette='plasma')
plt.title('Survival Rate by Number of Parents/Children Aboard')
plt.ylabel('Survival Rate')
plt.show()

# 6. Survival by Fare (Distribution)
plt.figure(figsize=(10, 6))
sns.histplot(df_titanic[df_titanic['Survived'] == 1]['Fare'], color='green', kde=True, label='Survived', bins=30)
sns.histplot(df_titanic[df_titanic['Survived'] == 0]['Fare'], color='red', kde=True, label='Not Survived', bins=30)
plt.title('Fare Distribution of Survivors vs. Non-Survivors')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.legend()
plt.show()

# 7. Survival by Embarked (after one-hot encoding)
# Reconstruct Embarked categories for plotting if needed or plot directly with one-hot encoded columns
# For simplicity, let's plot the original Embarked if we had it, or use the dummy columns
# As we have df_titanic_encoded with Embarked_Q, Embarked_S, let's visualize these

# If you wanted to visualize the effect of Embarked on survival before dropping 'Embarked' and after imputation:
# You would typically group by Embarked and calculate mean Survived.
# Let's consider the effects of the one-hot encoded columns on 'Survived'
plt.figure(figsize=(8, 5))
sns.barplot(x='Embarked_Q', y='Survived', data=df_titanic, palette='rocket')
plt.title('Survival Rate for Embarked Q')
plt.ylabel('Survival Rate')
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(x='Embarked_S', y='Survived', data=df_titanic, palette='rocket')
plt.title('Survival Rate for Embarked S')
plt.ylabel('Survival Rate')
plt.show()

print("Visualizations completed for key features vs. Survival.")
```
